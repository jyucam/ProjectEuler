# Problem 9: Special Pythagorean Triplet

A Pythagorean triplet is a set of three natural numbers, $a<b<c$, for which, 
$$a^2 + b^2 = c^2$$
For example, $3^2 + 4^2 = 5^2$. 

There exists exactly one Pythagorean triple for which $a+b+c=1000$.
Find the product $abc$.

```python
c = 1   # Initialise pyth triplets
i = 0   # Initialise iterator

def pyth(x):
    """If x is a pythagorean number c, return a+b+c and a*b*c"""
    triples = []
    for i in range(x):
        for j in range(i):
            if i**2 + j**2 == x**2:
                triples.append((i+j+x, i*j*x))
    return triples

for c in range(500):
    for e in pyth(c):
        if e[0] == 1000:
            print(c, e)
            break
```
