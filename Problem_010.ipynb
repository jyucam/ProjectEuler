# Problem 10: Summation of Primes

The sum of the primes below $10$ is $2+3+5+7 = 17$.

Find the sum of all the primes below two million.



```python
def is_prime(x):
    """Checks if some integer x is prime"""

    # 1. Handle base cases 
    if x <= 1: 
        return False
    if x <= 3: 
        return True # 2 and 3 are prime

    # Exclude multiples of 2 and 3
    if x % 2 == 0 or x % 3 == 0:
        return False

    # Check 6k +/- 1 up to the square root of x
    i = 6   # initialise iterator 
    while i*i <= x:
        if x % (i-1) == 0 or x % (i+1) == 0:
            return False
        i += 6
    
    return True # Returns true if not divisible by any integer less than x

prime_sum = 0   #Initialise prime sum

for j in range(2,2000000):
    """Append i to sum if i is a prime number"""
    if is_prime(j):
        prime_sum += j

print(prime_sum)
```
