# Getting Divisors of a Number

For some integer $x$, the following algorithm checks if each number from $1$ to $\lfloor \sqrt{x} \rfloor$ is a divisor of $x$. If so, said number $i$ as well as $x/i$ are appended to a list of divisors. 

Then, if $\sqrt{x}$ is an integer, also adds $\sqrt{x}$ to the list of divisors. 

```python
def get_divisors(x):
    """Gets the divisors for an integer x in O(sqrt{x})"""
    divisors = []   # Initialise divisors
    i = 1           # Initialise iterator

    # Add i and x // i to divisors if i divides x
    while i**2 < x:
        if x % i == 0:
            divisors.append(i)
            divisors.append(x//i)
        i += 1
    if i**2 ==x:
        divisors.append(i)

    divisors.sort()

    return divisors
```

```python
get_divisors(120)
```
