# Checking if a number if prime

If we want to create an array of primes, check the page on the [Sieve of Eratosthenes](Sieve_of_Eratosthenes.ipynb). 

Otherwise, we can check if an integer $x$ is prime using the following algorithm. 

1. If $x \leq 1$, $x$ is not prime. 
1. If $x \leq 3$ (i.e. $x=2, 3$) $x$ is prime. 
1. If $x$ is a multiple of $2$ or $3$, $x$ is not prime. 
1. If $x$ is a multiple of $6n \pm 1$, $x$ is not prime. 

Consider the values of $6n+k$ where $k \in \mathbb{Z}$ and $0\leq k \leq 5$. 
1. If $k \in {0, 2, 4}$ then $6n+k$ is divisible by 2.
1. If $k = 3$, then $6n+k$ is divisible by 3.
1. $6n+5$ can be written as $6m-1$ where $m=n+1$. 

The first prime number greater than $3$ is $5 = 6(1)-1$. Therefore, all integers not a multiple of $2$ or $3$ can be written in the form $6n\pm 1$ where $n\in \mathbb{Z}$ and $n\geq 1$. 

```python
def is_prime(x):
    """Checks if some integer x is prime"""

    # 1. Handle base cases 
    if x <= 1: 
        return False
    if x <= 3: 
        return True # 2 and 3 are prime

    # Exclude multiples of 2 and 3
    if x % 2 == 0 or x % 3 == 0:
        return False

    # Check 6k +/- 1 up to the square root of x
    i = 6   # initialise iterator 
    while i*i <= x:
        if x % (i-1) == 0 or x % (i+1) == 0:
            return False
        i += 6
    
    return True # Returns true if not divisible by any integer less than x
```
