# Problem 26: Reciprocal Cycles

A unit fraction contains $1$ in the numerator. The decimal representation of the unit fractions with denominators $2$ to $10$ are given:

$$
    \begin{align*}
    1/2 &= 0.5\\
    1/3 &=0.(3)\\
    1/4 &=0.25\\
    1/5 &= 0.2\\
    1/6 &= 0.1(6)\\
    1/7 &= 0.(142857)\\
    1/8 &= 0.125\\
    1/9 &= 0.(1)\\
    1/10 &= 0.1
\end{align*}
$$

Where $0.1(6)$ means $0.166666\dots$, and has a $1$-digit recurring cycle. It can be seen that $1/7$ has a $6$-digit recurring cycle.

Find the value of $d < 1000$ for which $1/d$ contains the longest recurring cycle in its decimal fraction part.

What we will do is essentially a long division algorithm, where we input a prime into a function $\textrm{period}$ to obtain the period of the repeating decimal $1/p$. 

We will first initialise our divident, which is 1. Then, we compute the remainder of $1 \div p$ (given by $d = 1 \mod p$). We save this position of $d$. 

We will now multiply $d$ by $10$ (our base), and compute remainder of $d \div p$.

1. If $d = 0$, then the division terminates, so there is no repeating decimal. 
1. If $d$ is a value we've previously saved, the division algorithm will repeat. Thus, the length of the repeating decimal is the current position minus the position of $d$ saved. 
1. Else, save $(d, \textrm{pos}(d))$ to some sort of memory (here, a dictionary is used), then multiply $d$ by 10. 

```python
def prime_list(x):
    """Outputs a table of prime numbers less than or equal to x"""

    if x < 2:  # Return empty list if x <= 2 (no primes)
        return [] 

    a = [True for i in range(2, int(x)+1)] # Initialise list of truths (basically is_prime)
    p = []                          # Initialise list of primes

    i = 2
    while i <= x:
        # if a[i-2] is true, i.e. i is a prime number, append i to p (list of primes)
        if a[i-2] == True:
            p.append(i) 

            # Set all multiples of i greater than i**2 to be not prime
            # All integers less than i**2 which are not prime have already been accounted for
            j = i
            while j <= x // i:
                a[i*j - 2] = False
                j += 1

        i += 1
    
    return p # Returns list of primes 

primes = prime_list(1000)

print(len(primes))
```

```python
def period(p):
    """Determines the period of repeating decimals of prime p"""
    mem = {}
    d = 1       # Initialise divident 
    pos = 0     # Initialise position variable for dict

    while pos < 4e6:
        pos += 1    # Iterate position
        d = d % p   # Compute remainder
        if d == 0:      # No remainder, division terminates, no period
            return 0    
        elif d in mem:  # Same remainder / divident as before, division repeats, return period
            return pos - mem[d]
        else:           # Save position of remainder 
            mem[d] = pos
        d *= 10     # Calculate new divident (remainder * 10)

    return 0

max_length = 0
max_prime = 0

for p in primes:
    if period(p) > max_length:
        max_length = period(p)
        max_prime = p

print(max_prime, max_length)
```
