# Problem 37: Truncatable Primes

The number $3797$ has an interesting property. Being prime itself, it is possible to continuously remove digits from left to right, and remain prime at each stage: $3797$, $797$, $97$, and $7$. Similarly we can work from right to left: $3797$, $379$, $37$, and $3$.

Find the sum of the only eleven primes that are both truncatable from left to right and right to left.

NOTE: $2$, $3$, $5$, and $7$ are not considered to be truncatable primes.

First constraints I noticed is that the truncated primes must start with 2, 3, 5 or 7 and end with a 1, 3, 7 or 9. I have a feeling that there is a way to build such numbers. We know there is a maximum of 11 such numbers, but not that there is an upper bound to try. 

```python
def prime_list(x):
    """Outputs a table of prime numbers less than or equal to x"""

    if x < 2:  # Return empty list if x <= 2 (no primes)
        return [] 

    a = [True for i in range(2, int(x)+1)] # Initialise list of truths (basically is_prime)
    p = []                          # Initialise list of primes

    i = 2
    while i <= x:
        # if a[i-2] is true, i.e. i is a prime number, append i to p (list of primes)
        if a[i-2] == True:
            p.append(i) 

            # Set all multiples of i greater than i**2 to be not prime
            # All integers less than i**2 which are not prime have already been accounted for
            j = i
            while j <= x // i:
                a[i*j - 2] = False
                j += 1

        i += 1
    
    return p # Returns list of primes 

primes = prime_list(1000000)
primes_set = set(primes)

def is_prime(x):
    """Checks if some integer x is prime"""

    # 1. Handle base cases 
    if x <= 1: 
        return False
    if x <= 3: 
        return True # 2 and 3 are prime

    # Exclude multiples of 2 and 3
    if x % 2 == 0 or x % 3 == 0:
        return False

    # Check 6k +/- 1 up to the square root of x
    i = 6   # initialise iterator 
    while (i-1)**2 <= x:
        if x % (i-1) == 0 or x % (i+1) == 0:
            return False
        i += 6
    
    return True # Returns true if not divisible by any integer less than x
```

```python
import math

# Building the truncatable primes

truncatable = []
truncatable_count = 0
truncatable_prefix = [2, 3, 5, 7]

while truncatable_count < 11:
    """This loop builds a list of truncable primes as prefixes and appends an odd number to check if the new one is left-to-right truncatable"""

    truncatable_prefix_new = [] # Initialise list of next truncatable prefixes

    for i in truncatable_prefix:    # Builds list of primes from a truncatable prefix 
        for j in [1, 3, 7, 9]:
            new_prime = i * 10 + j
            # Check it is right hand side truncatable
            y = new_prime
            y_is_prime = True

            while y >= 10:
                if not is_prime(y):
                    y_is_prime = False
                    break
                n = math.floor(math.log10(y))
                a = y // (10**n)
                y -= a* 10**n

            if is_prime(new_prime):
                truncatable_prefix_new.append(new_prime)

            if y_is_prime and new_prime % 10 != 9 and new_prime % 10 != 1:
                truncatable.append(new_prime)
                truncatable_count += 1
                
    truncatable_prefix = truncatable_prefix_new

print(truncatable, sum(truncatable))
```
