# Problem 14: Longest Collatz Sequence

The following iterative sequence is defined for the set of positive integers:
- $n \to \frac{n}{2}$ ($n$ is even)
- $n \to 3n +1$ ($n$ is odd)

Using the rule above and starting with $13$, we generate the following sequence: 
$$13 \to 40 \to 20 \to 10 \to 5 \to 16 \to 8 \to 4 \to 2 \to 1$$

It can be seen that this sequence (starting at $13$ and finishing at $1$) contains $10$ terms. Although it has not been proved yet (Collatz Problem), it is thought that all starting numbers finish at $1$.

Which starting number, under one million, produces the longest chain?

**NOTE:** Once the chain starts the terms are allowed to go above one million.

Attempted 03/10/2026

```python
# We are going to assume Collatz sequences finish at 1

def next_collatz(x):
    """Returns the next value of the collatz sequence"""
    if int(x) % 2 == 0:
        return int(x/2)
    else:
        return int(3*x + 1)

longest_chain = (0, 0)

for i in range(1, 1000000):
    """Iterates from 1 to 999999 to find the length of the Collatz sequence"""
    """Updates longest chain when length exceeds previous longest chain"""
    j = 1   # Initialise iterator / length (1 term)
    x = i   # Initialise Collatz starting number

    while x > 1 and j < 4e6:    # If x > 1 (limited to 4e6 operations)
        x = next_collatz(x)     # Get next value in Collatz sequence
        j += 1                  # Iterate iterator by 1

    if j > longest_chain[1]:    # Update longest chain when j exceeds previous longest length
        longest_chain = (i, j)

print(longest_chain)
```

## A More Efficient Method

1. Many values are being recalculated, for instance: $\textrm{length}(26) = \textrm{length}(13) + 1$. It's much faster to store $\textrm{length}(13) = 10$ and calculate $\textrm{length}(26) = 11$. 
1. If $n$ is even, then $n \to \frac{n}{2}$, which means $\textrm{length}(2k) > \textrm{length}(k)$ for all $k \in \mathbb{Z}^+$. Therefore, we do not need to compute the chain for any $k < 500,000$.
1. If $n$ is odd, then $3n + 1$ is even, so $n \to 3n+1 \to \frac{3n+1}{2}$. We can save a step by giving $\textrm{length}(n) = \textrm{length}(\frac{3n+1}{2}) + 2$. 

Implementing this will give:

```python
values = {1: 1} # Initialise dictionary of lengths, assumes Collatz sequence ends at 1 (1 term)

def count_chain(x, v = values):
    """Computes the length of a Collatz sequence with starting number x"""
    """Instead of just the Collatz recursion, we turn calculating the length into a recursion too"""
    """We store the value of the length so it doesn't have to be recalculated"""

    if x in v.keys():
        return v[x] # Return v[x] if we already know the length 
    
    if x % 2 == 0:
        v.update({x: 1 + count_chain(x//2)})        # Stores v[x] = [x/2] + 1 if x is even (as above)
    else:
        v.update({x: 2 + count_chain((3*x +1)//2)}) # Stores v[x] = v[(3x+1)/2] if x is odd (as above)

    return v[x] # Return v[x] now that we know the length 

longest_chain = (0, 0)

for i in range(500000, 1000000):
    if count_chain(i) > longest_chain[1]:
        longest_chain = (i, count_chain(i))

print(longest_chain)
```
