# Problem 3: Largest Prime Factor

The prime factors of $13195$ are $5, 7, 13$ and $29$.

What is the largest prime factor of the number $600851475143$?

Attempted 29/09/2026

```python
n = 600851475143    # Initialise number

def is_prime(x):
    """Checks if some integer x is prime"""
    for i in range(2, x):
        if x % i == 0:
            return False # Returns false if divisible by any integer
    
    return True # Returns true if not divisible by any integer less than x

def largest_prime_factor(x):
    """Prints the largest prime factor, fast"""
    factor = x  # Initialise largest factor

    for i in range(2, x):
        if is_prime(i): # Checks if i is prime

            # Divides factor by i until it it no longer divides
            while factor % i == 0 and i <= factor: 
                factor = factor // i
            
            # If the new factor is prime, return this factor 
            if is_prime(factor):
                return factor 

print(largest_prime_factor(n))
```
