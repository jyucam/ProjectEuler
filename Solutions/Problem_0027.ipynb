# Problem 27: Quadratic Primes

Euler discovered the remarkable quadratic formula:
$$n^2 + n + 41$$
It turns out that the formula will produce $40$ primes for the consecutive integer values $0 \leq n \leq 29$. However, when $n=40$, $40^2 + 40 + 41 = 40(40+1) + 41$ is divisible by $41$, and certain when $n = 41, 41^2 + 41 + 41$ is clearly divisible by $41$. 

The incredible formula $n^2 - 79n + 1601$ was discovered, which produces $80$ primes for the consecutive values $0 \leq n \leq 79$. The product of the coefficients, $-79$ and $1601$, is $-126479$. 

Considering quadratics of the form: 

$n^2 + an + b$, where $|a| \lt 1000$ and $|b| \leq 1000$, where $|n|$ is the modulus/absolute value of $n$

Find the product of the coefficients, $a$ and $b$, for the quadratic expression that produces the maximum number of primes for consecutive values of $n$, starting with $n=0$.

Let our quadratic be $f(n) = n^2 + an + b$. Considering we want to start with a maximum number of consecutive primes, starting from $n=0$, $f(0)=b$, so $b$ must be a prime number. 

Additionally, when $n=b$, $f(b) = b^2 + ab + b$ is divisible by $b$. Hence, the maximum number of primes for consecutive values of $n \in \mathbb{Z}, n \geq 0$ is $b$. 

We will assume that there are values of $n$ where $f(n)$ produces more than 80 primes, so we set the value of $b$ to be $79 \leq b \leq 1000$ where $b$ is prime. 

Any prime number greater than $2$ is odd. For $f(n)$ to be odd, $n^2 + an$ must be even (as odd + even = odd). When $n$ is even, $n^2 + an$ is always even. When $n$ is odd, $n^2 + an = n(n+a)$. If $a$ is $\pm$ even, $n+a$ is odd and so $n(n+a)$ is odd. Therefore, $a$ must be $\pm$ odd. Hence, $a$ will be comprised of positive and negative versions of odd integers less than 1000. 

```python
# Create list of a
a_list = []
for i in range(500):
    a_list.append(2*i + 1)
    a_list.append(-(2*i + 1))
a_list.sort()
print(a_list[0:5], len(a_list))

def prime_list(x, y):
    """Outputs a table of prime numbers between x and y inclusive"""

    if y < 2:  # Return empty list if x <= 2 (no primes)
        return [] 

    a = [True for _ in range(2, int(y)+1)] # Initialise list of truths (basically is_prime)
    p = []                          # Initialise list of primes

    i = 2
    while i <= y:
        # if a[i-2] is true, i.e. i is a prime number, append i to p (list of primes)
        if a[i-2] == True:
            p.append(i) 

            # Set all multiples of i greater than i**2 to be not prime
            # All integers less than i**2 which are not prime have already been accounted for
            j = i
            while j <= y // i:
                a[i*j - 2] = False
                j += 1

        i += 1

    k = 0
    while p[k] < x:
        # if p[k] is less than x, iterate k
        k += 1
    
    return p[k:]    # Returns list of primes 

# Create list of b
b_list = prime_list(79, 1000)
print(b_list[0:5], len(b_list))
```

```python
def is_prime(x):
    """Checks if some integer x is prime"""

    # 1. Handle base cases 
    if x <= 1: 
        return False
    if x <= 3: 
        return True # 2 and 3 are prime

    # Exclude multiples of 2 and 3
    if x % 2 == 0 or x % 3 == 0:
        return False

    # Check 6k +/- 1 up to the square root of x
    i = 6   # initialise iterator 
    while (i-1)**2 <= x:
        if x % (i-1) == 0 or x % (i+1) == 0:
            return False
        i += 6
    
    return True # Returns true if not divisible by any integer less than x
```

```python
def max_n(a, b):
    """Returns the max value of n for consecutive n where f is prime"""
    n = 0   # Initialise n
    f = b   # Initialise f(0)
    while n < 4e6:
        f = n**2 + a*n + b  # Compute f(n)
        if is_prime(f):
            n += 1          # Iterate n if f is prime
        else: 
            return n        # Return n if f is not prime
    return False

print(max_n(1, 41))

total_max_n = 0 # Initialise total max n
a_b_pdt = 1     # Initialise product of coefficients

for a in a_list:
    for b in b_list:
        """Returns the total max n and the product of coefficients"""
        if max_n(a, b) > total_max_n:
            total_max_n = max_n(a, b)
            a_b_pdt = a*b

print(total_max_n, a_b_pdt)
```
