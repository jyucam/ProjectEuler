# Problem 4: Largest Palindrome Product

A palindromic number reads the same both ways. The largest palindrome made from the product of two
-digit numbers is $9009 = 91\times 99$.

Find the lagest palindrome made from the product of two 3-digit integers. 

Attempted 29/09/2026

```python
def is_palindrome_str(x):
    """Checks if some number x is a palindrome"""
    x_string = str(x)       # Obtain string of number x 
    
    for i in range(len(x_string)//2):
        # Check if ith digit = (length-i)th digit
        if x_string[i] != x_string[-i-1]: 
            return False    # Return false if not a palindrome
    
    return True # Return true if is a palindrome 
```

```python
palin_list = []

for i in range(899):
    for j in range(i, 899):
        if is_palindrome_str((999-i)*(999-j)):
            palin_list.append((999-i)*(999-j))
            break 

print(max(palin_list))
```

## A More Efficient Method

The following is code to test whether a number is a palindrome without using string methods (computationally similar but it's less cheaty if I don't use string methods)

```python
def reverse(x):
    """Uses non-string methods to return the reverse of a number"""
    reversed = 0
    while x > 0:
        reversed = 10*reversed + x % 10 # Takes last digit and multiplies by 10 each time
        x = x // 10                     # Removes last digit from x via integer divisiion
    return reversed 

def is_palindrome(x):
    """Uses the reverse function to test if number is palindrome"""
    return x == reverse(x)
```

Consider the largest palindrome that is a product of two 3-digit numbers. 

Because the palindrome $111111 = 143\times 777$ and the largest product of two 3-digit numbers is $998001 = 999\times 999$, we know the palindrome must be 6 digits long. 

Therefore, $P = \mathtt{xyzzyx} = 100001x + 10010y + 1100z = 11(9091x + 910y + 100z)$. 

Let $a$ and $b$ be two 3-digit numbers, such that $a \leq b$ and $P=ab$. Because $11$ is prime, at least one of $a$ or $b$ must be a multiple of $11$. 

```python
largest_palin = 0
a = 999

while a >= 100:
    """Checks for values of a and b, where at least 1 is a multiple of 11, where a*b is a palindrome"""
    """The added condition of a or b being a multiple of 11 means we check less values"""
    
    # Set the value of b and the value of db, the step size of the iterator 
    if a % 11 == 0:
        b = 999 # a is a multiple of 11, b does not have to be
        db = 1
    else:
        b = 990 # The largest multiple of 11 less than or equal to 999
        db = 11

    while b >= a:
        # Checks if a*b <= largest_palin
        # For a given a, no more values of b can produce a number greater than largest_palin
        if a*b <= largest_palin:
            break
        
        # Checks if a*b is a palindrome, if so update largest_palin 
        if is_palindrome(a*b):
            largest_palin = a*b
            
        b -= db # Iterates b
    a -= 1      # Iterates a

print(largest_palin)
    
```
