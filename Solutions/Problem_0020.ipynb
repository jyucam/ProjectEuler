# Problem 20: Factorial Digit Sum

$n!$ means $n\times (n-1) \imtes \dots \times 3 \times 2 \times 1$.

For example, $10! = 10\times 9 \times \dots \times 3 \times 2 \times 1 = 3628800$, \
and the sum of the digits in the number $10!$ is $3 + 6 + 2 + 8 + 8 + 0 + 0 = 27$.

Find the sum of the digits in the number $100!$.

Attempted 08/10/2026

```python
def factorial(x):
    """Returns the factorial of integer x"""
    f = 1
    for i in range(1, x+1):
        f *= i
    return f

def sum_digits(x):
    """Sums the digits in integer x"""
    y = x   # Initialise variable
    s = 0   # Initialise sum
    while y >= 1:
        s += y % 10 # Add last digit of y to s
        y = y // 10 # Remove last digit from y
    return s

print(sum_digits(factorial(100)))
```
