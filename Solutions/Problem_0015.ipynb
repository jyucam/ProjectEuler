# Problem 15: Lattice Paths

Starting in the top left corner of a $2\times 2$ grid, and only being able to move to the right and down, there are exactly $6$ routes to the bottom right corner.



How many such routes are there through a $20\times 20$ grid?

Attempted 04/10/2026

In a $20\times 20$ grid, where you may only move to the right or down, there must be $40$ total moves, $20$ of which will be to the right, and the other $20$ of which will be down. 

Therefore, of 40 total moves, choose 20 to be down, i.e. $\binom{40}{20} = \frac{40!}{20!(40-20)!}$

```python
def nCr(n, r):
    """This function returns nCr for integers n and r"""
    numerator = 1   # Initialise numerator
    denominator = 1 # Initialise denominator
    
    for i in range(n-(r-1), n+1):   # Computes numerator
        numerator *= i              # (n-(r-1))(n-(r-2))...(n-1)n

    for j in range(1, r+1):         # Computes denominator
        denominator *= j            # 1*2...(r-1)r

    return numerator // denominator

print(nCr(40, 20))
```
