# Problem 31: Coin Sums

In the United Kingdom the currency is made up of pound (£) and pence (p). There are eight coins in general circulation:

    1p, 2p, 5p, 10p, 20p, 50p, £1 (100p), and £2 (200p).

It is possible to make £2 in the following way:

    1×£1 + 1×50p + 2×20p + 1×5p + 1×2p + 3×1p

How many different ways can £2 be made using any number of coins?

Attempt 19/10/2026

---

In a more mathematical way, this means using the numbers 1, 2, 5, 10, 20, 50, 100 and 200, how many ways can you make the number 200. The first solution is trivial, which is just 1 £2 coin, i.e. $200 = 200$. 

Let $100a + 50b + 20c + 10d + 5e + 2f + g = 200$. Then iterate.

This feels like a really slow way to iterate, but I imagine it's fine. 

```python
count = 0

for a in range(3):
    if 100*a > 200:
        break
    for b in range(5):
        if 100*a + 50*b > 200:
            break
        for c in range(11):
            if 100*a + 50*b + 20*c > 200:
                break
            for d in range(21):
                if 100*a + 50*b + 20*c + 10*d > 200:
                    break
                for e in range(41):
                    if 100*a + 50*b + 20*c + 10*d + 5*e > 200:
                        break
                    for f in range(101):
                        if 100*a + 50*b + 20*c + 10*d + 5*e + 2*f > 200:
                            break
                        for g in range(201):
                            if 100*a + 50*b + 20*c + 10*d + 5*e + 2*f + g > 200:
                                break
                            if 100*a + 50*b + 20*c + 10*d + 5*e + 2*f + g == 200:
                                count += 1

print(count + 1)
```

# A More Efficient Method

Suppose we have an array of coins: `coins = [1, 2, 5, 10, 20, 50, 100, 200]`.

We can compute the number of ways we can make the $i$ th value using the coins in the array. 

Consider $c = 5$ and the value $i = 8$. The number of ways to make $8$ including the $c=5$ coin is the number of ways you can make a value of $8 - 5 = 3$. Therefore, you add $\textrm{ways}[3]$ to $\textrm{ways}[8]$. 

Another example would be $c=5$, $i=20$, you compute $\textrm{ways}[20] = \textrm{ways}[20] + \textrm{ways}[15]$. The value of $\textrm{ways}[15]$ already takes into account the coin $c=5$ as the loop iterates per list item. This leads to a much faster solution from 700 ms to 3 ms. 

```python
coins = [1, 2, 5, 10, 20, 50, 100, 200]

ways = [0 for _ in range(201)]  # The index ways[i] indicates the number of ways
                                # you can make the value i out of the coins.
ways[0] = 1                     # There is 1 way to make the value 0, no coins

for c in coins:
    for i in range(c, 201):
        ways[i] += ways[i - c]

print(ways[200])
```
