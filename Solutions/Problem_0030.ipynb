# Problem 30: Digit Fifth Powers

Surprisingly there are only three numbers that can be written as the sum of fourth powers of their digits: 

$$
\begin{align*}
1634 &= 1^4 + 6^4 + 3^4 + 4^4\\
8208 &= 8^4 + 2^4 + 0^4 + 8^4\\
9474 &= 9^4 + 4^4 + 7^4 + 4^4
\end{align*}
$$

As $1 = 1^4$ is not a sum it is not included. 

The sum of these numbers is $1634 + 8208 + 9474 = 19316$.

Find the sum of all the numbers that can be written as the sum of fifth powers of their digits.

Considering $1^5$ is not a sum, the minimum number of digits these numbers may have is $2^5 = 32$, so 2 digits. Let us also consider an upper bound. $9^5 = 59049$ has five digits. Therefore, multiplying this out by $5$ gives $295245$, which should act as an upper bound for numbers that we need to test. 

```python
fifth_power_sum_list = []   # Initialise list of power sums

def fifth_power_sum(x):
    """Checks if a number can be written as the sum of its digits to the fifth power"""
    s = 0   # Initialise sum
    y = x   # Initialise variable
    while y >= 1:
        s += (y % 10)**5    # Add last digit of y to the fifth power to s
        y = y // 10         # Remove last digit from y

    # Checks if the sum of the digits to the fifth power = the original number
    if s == x:
        return True
    else:
        return False

for i in range(32, 295245):
    if fifth_power_sum(i):
        fifth_power_sum_list.append(i)

print(fifth_power_sum_list, sum(fifth_power_sum_list))
```
