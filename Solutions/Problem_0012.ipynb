# Problem 12: Highly Divisible Triangular Number

The sequence of triangle numbers is generated by adding the natural numbers. So the $7$ th triangle number would be $1+2+3+4+5+6+7 = 28$. The first ten terms would be: 
$$1, 3, 6, 10, 15, 21, 28, 36, 45, 55$$
Let us list the factors of the first seven triangle numbers: 
$$
\begin{align*}
\mathbf 1 &\colon 1\\
\mathbf 3 &\colon 1,3\\
\mathbf 6 &\colon 1,2,3,6\\
\mathbf{10} &\colon 1,2,5,10\\
\mathbf{15} &\colon 1,3,5,15\\
\mathbf{21} &\colon 1,3,7,21\\
\mathbf{28} &\colon 1,2,4,7,14,28
\end{align*}
$$

We can see that $28$ is the first triangle number to have over five divisors.

What is the value of the first triangle number to have over five hundred divisors?

Attempted 01/10/2026

```python
def tri(x):
    """Computes the xth triangular number"""
    return int(x*(x+1)/2)

def count_factors(x):
    """Counts the number of factors of some integer x"""
    j = 1
    factors = 0

    while j**2 <= x:
        if x % j == 0:
            if j**2 == x:   
                factors += 1    # If x is square, count the factor when j**2 = x
            else:
                factors += 2    # Factors always come in pairs if x is not square
        j += 1
    return factors

i = 0 
while i < 4e6:
    """Iterates to find the first triangular number with over 500 factors"""
    t = tri(i)
    if count_factors(t) > 500:
        print(i, t, count_factors(t))
        break
    i += 1
```

## A More Efficient Method

The $n$th triangular number is given by $t_n = \frac{n(n+1)}{2}$. $n$ and $n+1$ are necessarily coprime, as $(n+1)-n = 1$ is the greatest common multiple they can have. The total number of divisors of $t_n$ can be obtained by:

$$
\begin{cases}
D(t_n) = D(\frac{n}{2})D(n+1) \qquad \textrm{if $n$ is even} \\ 
D(t_n) = D(n)D(\frac{n+1}{2}) \qquad \textrm{if $n+1$ is even}
\end{cases}

```python
# Generate an array of prime number, see Sieve of Eratosthenes
def prime_list(x):
    """Outputs a table of prime numbers less than or equal to x"""

    if x < 2:  # Return empty list if x <= 2 (no primes)
        return [] 

    a = [True for i in range(2, int(x)+1)] # Initialise list of truths (basically is_prime)
    p = []                          # Initialise list of primes

    i = 2
    while i <= x:
        # if a[i-2] is true, i.e. i is a prime number, append i to p (list of primes)
        if a[i-2] == True:
            p.append(i) 

            # Set all multiples of i greater than i**2 to be not prime
            # All integers less than i**2 which are not prime have already been accounted for
            j = i
            while j <= x // i:
                a[i*j - 2] = False
                j += 1

        i += 1
    
    return p # Returns list of primes 

primes = prime_list(1000)
print(primes)
```

```python
def divisor_count(x, p = primes):
    """Returns the count of divisors for a given integer x"""
    D = 1
    if x in p:  # If x is prime, return 2 divisors
        D = 2
    else:       # If x is not prime, check each prime for number of divisors
        y = x   # Initialise some number y
        for e in p:      
            exponent = 1        # Initialise exponent
            while y % e == 0:   # If y is divisible by prime e
                exponent += 1   # Increase exponent by 1, divide y by e
                y = y//e
            D = D*exponent

    return D

print(divisor_count(60))
```

```python
i = 1
while i < 4e6:
    """Iterates to find the first triangular number with over 500 factors"""
    if i % 2 == 0:
        if divisor_count(i//2)*divisor_count(i+1) > 500:
            print(i, tri(i), divisor_count(i//2)*divisor_count(i+1))
            break
    else:
        if divisor_count(i)*divisor_count((i+1)//2) > 500:
            print(i, tri(i), divisor_count(i)*divisor_count((i+1)//2))
            break
    i += 1
```
