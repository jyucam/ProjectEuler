# Problem 19: Counting Sundays

You are given the following information, but you may prefer to do some research for yourself.

- 1 Jan 1900 was a Monday.
- Thirty days has September,
    April, June and November.
    All the rest have thirty-one,
    Saving February alone,
    Which has twenty-eight, rain or shine.
    And on leap years, twenty-nine.
- A leap year occurs on any year evenly divisible by 4, but not on a century unless it is divisible by 400.

How many Sundays fell on the first of the month during the twentieth century (1 Jan 1901 to 31 Dec 2000)?

Attempted 08/10/2026

```python
d = 2   # Initialise day of the week (1 Jan 1901 was on a Tuesday)
s = 0   # Initialise no. of Sundays

for year in range(1901, 2001):
    for month in range(1, 13):
        """Counts the number of Sundays in each month from year 1901 to 2000"""
        
        # Add count to sundays if day of the week is 0 (Sunday)
        if d == 0:
            s += 1

        # Calculate day of the week of first day in 30 day months
        if month in [4, 6, 9, 11]:
            d = (d + 30) % 7    
        # Calculate day of the week of first day in February
        elif month == 2:
            if year % 4 == 0:
                d = (d + 29) % 7    # February has 29 days in a leap year
            else:
                d = (d + 28) % 7
        # Calculate day of the week of first day in 31 day months
        else:
            d = (d + 31) % 7

print(s)
```
