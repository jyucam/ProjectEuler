# Problem 38: Pandigital Multiples

Take the number $192$ and multiply it by each of $1$, $2$, and $3$:

$$
\begin{align*}
192 \times 1 &= 192\\
192 \times 2 &= 384\\
192 \times 3 &= 576
\end{align*}
$$

By concatenating each product we get the $1$ to $9$ pandigital, $192384576$. We will call $192384576$ the concatenated product of 192$ and $(1, 2, 3)$.

The same can be achieved by starting with $9$ and multiplying by $1$, $2$, $3$, $4$, and $5$, giving the pandigital, $918273645$, which is the concatenated product of $9$ and $(1, 2, 3, 4, 5)$.

What is the largest $1$ to $9$ pandigital $9$-digit number that can be formed as the concatenated product of an integer with $(1, 2, \dots n)$ where $n>1$?

To get the largest 1 to 9 pandigital number, we know that the number must at least start with a 9, as $9$ itself gives this result. Therefore, we only need to check numbers that begin with a 9. We also know that $n>1$, which means that the products must be of $5$-digits or shorter. For example, $5000$ and $5000\times 2$ gives $500010000$. Hence, we our list of numbers we need to try to produce the pandigital multiples must contain numbers smaller than $10^4$. 

```python
import math
nines_list = []
i = 1
while i < 1e4:
    if i // 10**math.floor(math.log10(i)) != 9:
        i = 9*10**math.floor(math.log10(i))
    nines_list.append(i)
    i += 1
print(nines_list[0:5], len(nines_list))
```

```python
def get_digits(x):
    """Gets the digits of an integer i"""
    y = x
    digits = []
    while y >= 1:
        digits.append(y % 10)
        y //= 10
    digits.reverse()
    return digits

def build_number(a):
    """Builds integer from array a"""
    res = a[0]
    for i in range(1, len(a)):
        res = res*10 + a[i]
    return res

max_value = 0

for x in nines_list:
    j = 1
    pdt = x
    digits = []
    skip = False
    while len(digits) < 9:
        digits.extend(get_digits(pdt))
        if len(digits) > len(set(digits)) or 0 in digits or len(digits) > 9:
            skip = True
            break
        j += 1
        pdt = x*j
    if skip == False:
        if build_number(digits) > max_value:
            max_value = build_number(digits)
            print(max_value, x)
    
```
