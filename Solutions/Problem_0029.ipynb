# Problem 29: Distinct Powers

Consider all integer combinations of $a^b$ for $2 \leq a \leq 5$ and $2 \leq b \leq 5$: 

$$
\begin{array}{rrrr}
2^2=4, &2^3=8, &2^4=16, &2^5=32\\
3^2=9, &3^3=27, &3^4=81, &3^5=243\\
4^2=16, &4^3=64, &4^4=256, &4^5=1024\\
5^2=25, &5^3=125, &5^4=625, &5^5=3125
\end{array}
$$

If they are then placed in numerical order, with any repeats removed, we get the following sequence of $15$ distinct terms: 

$$4, 8, 9, 16, 25, 27, 32, 64, 81, 125, 243, 256, 625, 1024, 3125$$

How many distinct terms are in the sequence generated by $a^b$ for $2 \leq a \leq 100$ and $2 \leq b \leq 100$?

A trivial solution would be to just compute $a^b$, iterating over all values of $a$ and $b$ and place them into a set to remove duplicates. It also already happens to be really fast. 

Another way to attempt the problem would be to recognise when a number $A$ is in the form $a^b$, for which more numbers up to $a^{100}$ will have already been saved. 

Therefore, we can skip some computation by setting the iterator $B$ to $\lfloor \frac{100}{b} \rfloor + 1$ and record all values of $A^B$ up to $B = 100$. 

As it turns out, this barely saved any time and will have also used more storage, as in addition to computing $a^b$, we also save the values of $a$ and $b$, effectively tripling storage. 

```python
powers_set = set()  # Initialise set of powers

for a in range(2, 101):
    for b in range(2, 101):
        """Iterates through a and b and adds each distinct power to a set"""
        powers_set.add(a**b)

print(list(powers_set)[0:5], len(powers_set))
```

```python
a = 2               # Initialise base iterator
powers_dict = {}    # Initialise powers dictionary

while a <= 100:
    b = 2           # Initialise power iterator
    if a in powers_dict.keys():
        b = 100 // powers_dict[a][1] + 1    # Jump to least value of a^b not already recorded
    while b <= 100:
        powers_dict.update({a**b: (a, b)})
        b += 1

    a += 1

print(len(powers_dict))
```
