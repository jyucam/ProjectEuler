# Problem 41: Pandigital Prime

We shall say that an $n$-digit number is pandigital if it makes use of all the digits $1$ to $n$ exactly once. For example, $2143$ is a $4$-digit pandigital and is also prime.

What is the largest $n$-digit pandigital prime that exists?

We should check the divisibility of numbers containing digits 1 through 9. 

Because $1 + 2 + 3 + 4 + 5 + 6 + 7 + 8 + 9 = 45$, therefore, any $1$ to $9$ pandigital number is divisible by $3$. 

Likewise, because $1 + 2 + 3 + 4 + 5 + 6 + 7 + 8 = 36$, therefore, any $1$ to $8$ pandigital number is divisible by $3$. 

Let's try $7$. Because $1 + 2 + 3 + 4 + 5 + 6 + 7 = 28$ is not a multiple of $3$, therefore, any $1$ to $7$ pandigital number is **not** divisible by $3$.

```python
def is_prime(x):
    """Checks if some integer x is prime"""

    # 1. Handle base cases 
    if x <= 1: 
        return False
    if x <= 3: 
        return True # 2 and 3 are prime

    # Exclude multiples of 2 and 3
    if x % 2 == 0 or x % 3 == 0:
        return False

    # Check 6k +/- 1 up to the square root of x
    i = 6   # initialise iterator 
    while (i-1)**2 <= x:
        if x % (i-1) == 0 or x % (i+1) == 0:
            return False
        i += 6
    
    return True # Returns true if not divisible by any integer less than x

def int_to_list(x):
    """Converts an integer x into a list of digits"""
    a = []  # Initialise list
    y = x   # Initialise variable
    while y >= 1:
        a.append(y % 10)
        y //= 10
    a[:] = reversed(a)
    return a

def list_to_int(a):
    """Converts a list a into an integer with digits of a"""
    res = 0 # Initialise result
    for i in range(len(a)):     # Return the list in number form 
        res = 10*res + a[i]
    return res

def next_permutation(x):
    """Computes the next lexicographic permutation of a number x"""    
    a = int_to_list(x)
    n = len(a)   # Initialise length of list

    # Find the position of the rightmost character which is less than its next character
    i = n - 2
    while i >= 0 and a[i] <= a[i+1]:
        i -= 1
    if i < 0:
        return False

    # Find the position of fist value to the right of i greater than i
    j = n - 1
    while j > i and a[j] >= a[i]:
        j -= 1

    a[i], a[j] = a[j], a[i] # Swap the found characters

    a[i+1:] = reversed(a[i+1:])   # Sort (reverse) all characters after i

    return list_to_int(a)
```

```python
n = 7654321
prime = False

while prime == False and next_permutation(n) != False:
    if is_prime(n):
        prime = True
    else:
        n = next_permutation(n)

print(n, prime)
```
