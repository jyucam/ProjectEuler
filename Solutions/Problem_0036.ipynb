# Problem 36: Double-base Palindromes

The decimal number, $585 = 1001001001_2$ (binary), is palindromic in both bases.

Find the sum of all numbers, less than one million, which are palindromic in base $10$ and base $2$.

(Please note that the palindromic number, in either base, may not include leading zeros.)

Attempted 24/10/2026

```python
def to_binary(x):
    """Converts a number to binary"""
    y = x   # Initialise variable
    i = 0   # Initialise order iterator
    b = 0   # Initialise binary result
    while y >= 1:
        b += y % 2 * 10**i
        i += 1
        y //= 2
    return b

print(to_binary(4))
print(to_binary(18))
print(to_binary(585))
```

```python
def reverse(x):
    """Reverses an integer x"""
    y = x   # Initialise variable
    r = 0   # Initialise result
    while y >= 1:
        r = 10*r + y % 10
        y //= 10
    return r

print(reverse(1002))
```

```python
def is_palindrome(x):
    """Checks if an integer x is a palindrome"""
    return x == reverse(x)

# Create array of double palindromes 
double_palindromes = []
for i in range(1, 1000001):
    if is_palindrome(i):
        if is_palindrome(to_binary(i)):
            double_palindromes.append(i)

print(sum(double_palindromes))
```

# A More Efficient Method

A more efficient way to solve this problem would be to generate palindromes rather than checking each integer from 1 whether they are a palindrome. 

If we have the $3$-digit number $xyz$, it defines the $5$-digit palindrome $xyzyx$ and the $6$-digit palindrom $xyzzyx$. In this case, we would have to generate about 20,000 palindromes rather than checking $10^8$ numbers. 

This method runs in rouhgly 6 ms, which is pretty good. 

```python
def make_palindrome(x, base, oddlength):
    """Make a palindrome from defining integer x"""
    result = x  # Initialise result
    y = x       # Initialise variable
    if oddlength:
        y = y // base
    # Iterate to make rest of palindrome
    while y >= 1:
        result = result*base + y % base
        y //= base
    return result 

limit = 1e6 # Set limit
total = 0   # Initialise sum

# For even-digit palindromes
i = 1       # Initialise iterator
p = make_palindrome(i, 2, True)     # Initialise palindrome number
while p < limit:
    if is_palindrome(p):
        total += p
    i += 1
    p = make_palindrome(i, 2, True)

# For odd-digit palindromes
i = 1       # Initialise iterator
p = make_palindrome(i, 2, False)    # Initialise palindrome number
while p < limit:
    if is_palindrome(p):
        total += p
    i += 1
    p = make_palindrome(i, 2, False)

print(total)
```
