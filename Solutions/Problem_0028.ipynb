# Problem 28: Number Spiral Diagonals

Starting with the number $1$ and moving to the right in a clockwise direction a $5$ by $5$ spiral is formed as follows:

||||||
|----|----|----|----|----|
| **21** | 22 | 23 | 24 | **25** |
| 20 | **7**  | 8  | **9**  | 10 |
| 19 | 6  | **1**  | 2  | 11 |
| 18 | **5**  | 4  | **3**  | 12 |
| **17** | 16 | 15 | 14 | **13** |

It can be verified that the sum of the numbers on the diagonals is $101$.

What is the sum of the numbers on the diagonals in a $1001$ by $1001$ spiral formed in the same way?

Note the numbers in the upper right diagonal must be an odd square. This is because the size of squares are always an odd number by an odd number, aka an odd square (i.e. $1, 9, 25, 49 \dots$) and the upper right-hand corner is always filled last. 

For the first layer (i.e. the $1\times 1$ spiral), the sum of diagonals is $1$. For all other diagonals, the sum of diagonals will consist of $4$ terms. 

Consider the $n$ th layer where $n>1$. Each diagonal will be $n-1$ less than the next. For $n=5$, we have $21 = 25 - 4$, $17 = 21 - 4$ and $13 = 17 - 4$. This is an arithmetic series with common difference $n-1$ from $n^2$ to $n^2 - 3(n-1)$, where the sum can be given as 
$$\frac{4(2n^2 - 3(n-1))}{2} = 4n^2 - 6(n-1) = 4n^2 - 6n + 6$$

E.g. for $n=5$, $4(5)^2 - 6(5-1) = 76$. 

For a $1001\times 1001$ grid, $n=501$, so the sum of diagonals is:

```python
sum = 1
i = 3
while i <= 1001:
    sum += 4*i*i - 6*i + 6
    i += 2

print(sum)
```

We can theoretically use deterministic formulae to get the sum of odd squares and odd integers, which would give us an $\mathcal{O}(1)$ solution, but for the sake of simplicity, $\mathcal{O}(n)$ is good enough for me.
