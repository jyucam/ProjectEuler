# Problem 4: Largest Palindrome Product

A palindromic number reads the same both ways. The largest palindrome made from the product of two
-digit numbers is $9009 = 91\times 99$.

Find the lagest palindrome made from the product of two 3-digit integers. 

Attempted 29/09/2026

```python
def is_palindrome(x):
    """Checks if some number x is a palindrome"""
    x_string = str(x)       # Obtain string of number x 
    
    for i in range(len(x_string)//2):
        # Check if ith digit = (length-i)th digit
        if x_string[i] != x_string[-i-1]: 
            return False    # Return false if not a palindrome
    
    return True # Return true if is a palindrome 
```

```python
palin_list = []

for i in range(899):
    for j in range(899):
        if is_palindrome((999-i)*(999-j)):
            palin_list.append((999-i)*(999-j))
            break 

print(max(palin_list))
```
