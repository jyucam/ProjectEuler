# Problem 1: Multiples of 3 and 5

If we list all the natural numbers below $10$ that are multiples of $3$ or $5$, we get $3, 5, 6$ and $9$. The sum of these multiples is $23$.

Find the sum of all the multiples of $3$ or $5$ below $1000$.

Attempted 29/09/2026

```python
n = 1000    # number of entries
sum = 0     # initialise sum variable

for i in range(1, 1000):
    """sums up the multiples of 3 and 5"""
    """uses this method instead of jump counting to avoid double counting"""
    if i % 3 == 0:
        """adds i to sum if i is divisible by 3, i.e. remainder of i / 3 == 0"""
        sum += i
    elif i % 5 == 0:
        """adds i to sum if i is divisible by 5, i.e. remainder of i / 5 == 0"""
        sum += i
    
print(sum)
```

## A More Efficient Method

The sum of integers divisible by an integer $p$ from $1$ to $n$ is given by
$$ p \sum_{i=1}^{\lfloor n/p \rfloor} i = \frac{p \lfloor n/p \rfloor(\lfloor n/p \rfloor+1)}{2} $$

We can add up the integers divisible by 3 and integers divisible by 5, then subtract the numbers divisible by both 3 and 5 (i.e. divisible by 15) to avoid double counting. 

This will give a solution of $\mathcal{O}(1)$

```python
import math

def sum_divisible_by(p, n = 999):
    k = math.floor(n/p)
    return int(p * k * (k+1) / 2)

print(sum_divisible_by(3) + sum_divisible_by(5) - sum_divisible_by(15))
```
