# Problem 34: Digit Factorials

$145$ is a curious number, as $1! + 4! + 5! = 1 + 24 + 120 = 145$.

Find the sum of all numbers which are equal to the sum of the factorial of their digits.

**Note:** As $1! = 1$ and $2! = 2$ are not sums they are not included.

Because the prompt says to locate all of these numbers, it is worth coming up with an upper bound to test for. 

Consider a number $99\dots 9 = 9! + 9! + \dots 9!$

The number $99\dots 9$ has $n$ digits and must be greater than $10^{n-1}$. Therefore, we can solve the inequality $n(9!) > 10^{n-1}$. Solving gives a value of $\lceil n \rceil = 8$. The intersection between the two happens at $\lceil 2.69649\times10^{6} \rceil \approx 3 \times 10^6$

```python
factorial = [1, 1, 2, 6, 24, 120, 720, 5040, 40320, 362880]

def factorial_sum(x):
    """Finds the digit factorials of an integer x"""
    y = x
    s = 0
    while y >= 1 and s <= x:
        s += factorial[y % 10]
        y //= 10
    return x == s

total = 0

for i in range(10, 3000000):
    if factorial_sum(i):
        total += i

print(total)
```
