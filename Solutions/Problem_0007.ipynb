# Problem 7: 10 001st Prime

By listing the first six prime numbers: $2, 3, 5, 7, 11$, and $13$, we can see that the $6$ th prime is $13$.

What is the $10~001$ st prime number? 

Attempted 29/09/2026

```python
n = 10001   # Initialise number

def is_prime(x):
    """Checks if some integer x is prime"""
    for i in range(2, x):
        if x % i == 0:
            return False # Returns false if divisible by any integer
    
    return True # Returns true if not divisible by any integer less than x

i = 0   # Initialise iterators
j = 2

prime_list = [] # Initialise prime list

while i < n and j < n**2:
    """Checks if a number is prime, if so, appends to prime list"""
    if is_prime(j):
        prime_list.append(j)
        i += 1  # Iterate i when a prime number is appended to prime_list

    j += 1 # Iterate j for each operation

print(len(prime_list))
print(prime_list[-1])
        
```
