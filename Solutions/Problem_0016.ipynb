# Problem 16: Power Digit Sum

$2^{15} = 32768$ and the sum of its digits is $3 + 2 + 7 + 6 + 8 = 26$.

What is the sum of the digits of the number $2^{1000}$.

Attempted 05/10/2026

```python
a = 2**1000

def digit_sum(x):
    """Returns the digit sum of a number"""
    y = x   # Initialise changing variable
    s = 0   # Initialise sum
    
    while y >= 1:
        s += y % 10 # Add last digit to sum
        y //= 10    # Remove last digit from y
    return s

print(a)
print(digit_sum(a))
```
