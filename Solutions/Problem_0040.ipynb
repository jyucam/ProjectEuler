# Problem 40: Champernowne's Constant

An irrational decimal fraction is created by concatenating the positive integers:
$$0.12345678910{\color{red}\mathbf 1}112131415161718192021\cdots$$
It can be seen that the $12$ th digit of the fractional part is $1$.

If $d_n$ represents the $n$ th digit of the fractional part, find the value of the following expression.

$$d_1 \times d_{10} \times d_{100} \times d_{1000} \times d_{10000} \times d_{100000} \times d_{1000000}$$

Attempted 29/10/2026

```python
import math

def d(n):
    """Returns the nth digit of the fractional part of Champernowne's Constant"""
    i = 0   # Represents the ith digit
    x = 0   # Represents the xth number
    while i < n:    # Gets the first number with last index greater than the needed digit
        x += 1
        i += math.floor(math.log10(x)) + 1
    while i > n:    # Gets the needed digit from the above number
        x //= 10
        i -= 1
    return x % 10

print(d(1)*d(10)*d(100)*d(1000)*d(10000)*d(100000)*d(1000000))
```
