# Problem 17: Number Letter Counts

If the numbers $1$ to $5$ are written out in words: one, two, three, four, five, then there are $3 + 3 + 5 + 4 + 4 = 19$ letters used in total.

If all the numbers from $1$ to $1000$ (one thousand) inclusive were written out in words, how many letters would be used?

**NOTE:** Do not count spaces or hyphens. For example, $342$ (three hundred and forty-two) contains $23$ letters and $115$ (one hundred and fifteen) contains $20$ letters. The use of "and" when writing out numbers is in compliance with British usage.

```python
import math

# Define dictionary of names
dict = {
    1: "one",
    2: "two",
    3: "three",
    4: "four",
    5: "five",
    6: "six",
    7: "seven",
    8: "eight",
    9: "nine",
    10: "ten",
    11: "eleven",
    12: "twelve",
    13: "thirteen",
    14: "fourteen",
    15: "fifteen",
    16: "sixteen",
    17: "seventeen",
    18: "eighteen",
    19: "nineteen",
    20: "twenty",
    30: "thirty",
    40: "forty",
    50: "fifty",
    60: "sixty",
    70: "seventy",
    80: "eighty",
    90: "ninety"
}

def count_letters(x):
    """Counts the number of letters in the word form of integer x"""
    if x == 1000:   # Return len("onethousand") if x == 1000
        return len("onethousand")

    y = x       # Initialise variable
    length = 0  # Initialise length
    three_digit = False     # Initialise whether the number is at least 3 digits long 

    if y > 99: # Return hundreds length
        digit = math.floor(y / 100)
        three_digit = True
        length += len(dict[digit] + "hundred")
        y -= digit*100

    if y > 19:  # Return tens length
        digit = math.floor(y / 10)
        length += len(dict[digit*10])

        if three_digit: # If number is a three digit number, include "and"
            length += len("and")
            
        y -= digit*10
        three_digit = False

    if y > 0:   # Return unit length up to 19
        length += len(dict[y])

        if three_digit: # If number is a three digit number, include "and" (if not previously included)
            length += len("and")

    return length

total_length = 0

for i in range(1, 1001):
    total_length += count_letters(i)

print(total_length)
```
