# Problem 35: Circular Primes

The number, $197$, is called a circular prime because all rotations of the digits: $197$, $971$, and $719$, are themselves prime.

There are thirteen such primes below $100$: $2, 3, 5, 7, 11, 13, 17, 31, 37, 71, 73, 79$, and $97$.

How many circular primes are there below one million?

Attempted 23/10/2026

```python
def prime_list(x):
    """Outputs a table of prime numbers less than or equal to x"""

    if x < 2:  # Return empty list if x <= 2 (no primes)
        return [] 

    a = [True for i in range(2, int(x)+1)] # Initialise list of truths (basically is_prime)
    p = []                          # Initialise list of primes

    i = 2
    while i <= x:
        # if a[i-2] is true, i.e. i is a prime number, append i to p (list of primes)
        if a[i-2] == True:
            p.append(i) 

            # Set all multiples of i greater than i**2 to be not prime
            # All integers less than i**2 which are not prime have already been accounted for
            j = i
            while j <= x // i:
                a[i*j - 2] = False
                j += 1

        i += 1
    
    return p # Returns list of primes 

primes = prime_list(1000000)

print(primes[0:5], len(primes))
```

```python
def rotate(x):
    """Rotates a string by placing the first digit at the end"""
    return str(x)[1:] + str(x)[0]
    
print(rotate(197))
print(rotate(234))
print(rotate(21))
print(rotate(2))
print(rotate(101))
```

```python
prime_set = set(primes)
circular_primes = set()

for p in primes:
    if p not in circular_primes:
        q = rotate(p)
        circular = True
        rotations = set([p])

        while int(q) != p:
            if int(q) not in prime_set:
                circular = False
            rotations.add(int(q))
            q = rotate(q)

        if circular:
            circular_primes.update(rotations)
    
print(len(circular_primes))
```
