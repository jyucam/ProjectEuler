# Problem 33: Digit Cancelling Fractions

The fraction $49/98$ is a curious fraction, as an inexperienced mathematician in attempting to simplify it may incorrectly believe that $49/98 = 4/8$, which is correct, is obtained by cancelling the $9$ s.

We shall consider fractions like, $30/50 = 3/5$, to be trivial examples.

There are exactly four non-trivial examples of this type of fraction, less than one in value, and containing two digits in the numerator and denominator.

If the product of these four fractions is given in its lowest common terms, find the value of the denominator.

For two 2-digit numbers in a fraction, the fraction can be written as: 

$$\frac{10a + b}{10a + c}, \frac{10a + b}{10c + a}, \frac{10b + a}{10a + c}, \frac{10b + a}{10c + a}$$

, where $a$ is the repeating digit. In each case, we care looking for solutions where the above fraction is equal to $b/c$. 

```python
fractions = []  # Initialise list of fractions

for a in range(1, 10):
    for c in range(1, 10):
        for b in range(1, 10):
            """Determines if a fraction is digit cancelling"""
            # Check that b is not greater than or equal to c so b/c is less than 1
            if b >= c:
                break

            # Check if the fraction is digit cancelling. If so, append to fractions list
            if (10*a+b)/(10*a+c) == b/c:
                print((10*a+b), (10*a+c))
                fractions.append((b, c))

            if (10*a+b)/(10*c+a) == b/c:
                print((10*a+b),(10*c+a))
                fractions.append((b, c))

            if (10*b+a)/(10*a+c) == b/c:
                print((10*b+a), (10*a+c))
                fractions.append((b, c))

            if (10*b+a)/(10*c+a) == b/c:
                print((10*b+a), (10*c+a))
                fractions.append((b, c))

            # There are more efficient ways to write this code 
            # but I wanted to print the pairs to verify the results by inspection 

print(fractions)
```

So the fractions we end up with are:
$$\frac{1}{4}, \frac{2}{5}, \frac{1}{5}, \frac{4}{8}$$

Computing their product gives:
$$\frac{1}{4} \times \frac{2}{5} \times \frac{1}{5} \times \frac{4}{8} = \frac{1}{100}$$
