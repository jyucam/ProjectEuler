# Problem 5: Smallest Multiple

$2520$ is the smallest number that can be divided by each of the numbers from $1$ to $10$ without any remainder.

What is the smallest positive number that is **evenly divisible** (divisible with no remainder) by all of the numbers from $1$
to $20$?

```python
def is_prime(x):
    """Checks if some integer x is prime"""
    for i in range(2, x):
        if x % i == 0:
            return False # Returns false if divisible by any integer
    
    return True # Returns true if not divisible by any integer less than x

def prime_factorisation(x):
    """Prints the largest prime factor, fast"""
    factor = x          # Initialise largest factor
    factors_dict = {}   # Initialise factors dictionary 

    for i in range(2, x+1):
        if is_prime(i): # Checks if i is prime

            # Divides factor by i until it it no longer divides
            j = 0 # counter for each prime factor
            while factor % i == 0 and i <= factor:
                factor = factor // i
                j += 1
            
            factors_dict.update({i: j})

            # If the new factor is prime, return the factor list
            if is_prime(factor):
                factors_dict.update({factor: 1})
                return factors_dict 

def least_common_multiple(int_list):
    """Computes the least common multiple using prime factorisations"""
    f = {}  # Initialise factors dictionary 
    pdt = 1 # Initialise product

    # Get prime factorisation of the LCM 
    for e in int_list:
        p = prime_factorisation(e)
        keys = set(f.keys()).union(p.keys())    # Union of keys of p and f

        # for each k, f[k] = max(f[k], p[k])
        for k in keys:  
            if k in f.keys() and k in p.keys():
                f.update({k: max(f[k], p[k])})  
            elif k in p.keys():
                f.update({k: p[k]})

    # Compute the LCM
    for k in f.keys():
        pdt *= k**f[k]
    return pdt

factors = [i for i in range(2,21)]

print(factors)
print(least_common_multiple(factors))
```
